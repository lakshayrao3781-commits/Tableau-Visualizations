Q9. Chart Selection Justification
Explain why different charts were used for different business questions in this analysis.

Bar Chart (Category/Region): Ideal for comparing discrete categorical values clearly.   

Line Chart (Sales Trend & Cumulative Growth): Best suited for visualizing continuous time-series data and trends.  

 Scatter Plot (Sales vs Profit): Effective for highlighting correlations, outliers, and negative profit points.  

  Pie / Donut Chart (Regional Contribution): Helps show proportional composition relative to the total whole.
